```python
def plot_GSEA(condition, threshold=0.3, max_labels=None, eps=1e-300, cap_p=-np.log10(1e-300)):
    """
    eps: smallest FDR value to clip to (avoid inf)
    cap_p: plotting cap for -log10(FDR) if any inf slipped through (default consistent with eps)
    """
    res = pd.read_csv(os.path.join(results_dir, f'GSEA_{condition}.csv')).copy()
    res['-log10(FDR)'] = np.where(res['-log10(FDR)'] > 2, -np.log10(0.01), res['-log10(FDR)'])

    # --- Ensure needed columns exist ---
    # If '-log10(FDR)' is not present, compute from 'FDR'
    if '-log10(FDR)' not in res.columns:
        if 'FDR' not in res.columns:
            raise KeyError("Input must have either '-log10(FDR)' or 'FDR' column.")
        # Clip FDR into (eps,1] to avoid log10(0) and negatives
        fdr = pd.to_numeric(res['FDR'], errors='coerce')
        fdr = fdr.clip(lower=eps, upper=1.0)
        res['-log10(FDR)'] = -np.log10(fdr)

    # Coerce numeric columns
    res['NES'] = pd.to_numeric(res['NES'], errors='coerce')
    res['-log10(FDR)'] = pd.to_numeric(res['-log10(FDR)'], errors='coerce')

    # Replace +/-inf with a finite cap
    inf_mask = ~np.isfinite(res['-log10(FDR)'])
    if inf_mask.any():
        res.loc[inf_mask, '-log10(FDR)'] = cap_p

    # Drop rows with missing essentials
    needed = ['NES', '-log10(FDR)', 'Term']
    for col in needed:
        if col not in res.columns:
            raise KeyError(f"Missing required column: {col}")

    before = len(res)
    res = res.dropna(subset=needed)
    if len(res) == 0:
        raise ValueError("No finite data left after cleaning. Check your input file.")


    # --- Plotting ---
    res['NES_color'] = np.where(res['NES'] < 0, 'downregulated', 'upregulated')
    res['sig'] = res['-log10(FDR)'] > threshold

    lab_df = res[res['sig']].sort_values('-log10(FDR)', ascending=False)
    if max_labels is not None:
        lab_df = lab_df.head(max_labels)

    fig, ax = plt.subplots(figsize=(4, 3), dpi=300)

    scatter = sns.scatterplot(
        data=res, x='NES', y='-log10(FDR)',
        hue='NES',
        palette='coolwarm',
        edgecolor='black', s=100, legend=False, ax=ax
    )

    texts = []
    # initial small x-offset away from the point center
    for _, row in lab_df.iterrows():
        x0 = row['NES'] - (0.15 if row['NES'] > 0 else -0.15)
        texts.append(
            ax.annotate(
                row['Term'],
                xy=(row['NES'], row['-log10(FDR)']),
                xytext=(x0, row['-log10(FDR)']),
                textcoords='data',
                fontsize=7,
                ha='right' if row['NES'] > 0 else 'left',
                va='center',
                # bbox=dict(boxstyle='round,pad=0.15', fc='white', ec='none', alpha=0.8)  # optional
            )
        )

    ax.margins(x=0.1, y=0.12)

    adjust_text(
        texts,
        ax=ax,
        add_objects=scatter.collections,          # avoid points
        only_move={'points': 'y', 'texts': 'xy'},
        autoalign='y',
        expand_points=(1.2, 1.6),
        expand_text=(1.1, 1.3),
        force_points=0.2,
        force_text=0.5,
        lim=300,
        arrowprops=dict(arrowstyle='-', lw=0, color='0.4', shrinkA=2, shrinkB=2)
    )

    ax.set_xlabel('Normalized Enrichment Score (NES)', fontsize=12)
    ax.set_ylabel('-log10(FDR q-value)', fontsize=12)
    fig.tight_layout()
    return fig, ax

# Usage
fig, ax = plot_GSEA('Full McFarland dataset', threshold=0.8, max_labels=60)
plt.savefig(os.path.join(figures_dir, 'GSEA_McFarland.pdf'), bbox_inches='tight')
plt.show()
```

```python
def create_gsea_heatmap(dotplot_data, figure_size=(8, 4), legend=True):

    # Cluster rows
    nes_matrix = dotplot_data.pivot(index="Term", columns="model", values="NES").fillna(0)
    linkage_rows = linkage(pdist(nes_matrix, metric="cosine"), method="complete")
    row_order = nes_matrix.index[leaves_list(linkage_rows)].tolist()
    row_annotations = row_order.copy()

    print(row_order)

    from matplotlib.colors import PowerNorm

    # Color normalization
    max_nes = dotplot_data['NES'].abs().quantile(0.95)
    norm = PowerNorm(gamma=0.5, vmin=-max_nes, vmax=max_nes)

    # Normalize size
    min_size = 20
    max_size = 200
    size_vals = dotplot_data['-log10(FDR q-val)']
    dotplot_data['SizeNorm'] = (
        (dotplot_data['-log10(FDR q-val)'] - dotplot_data['-log10(FDR q-val)'].min()) /
        (dotplot_data['-log10(FDR q-val)'].max() - dotplot_data['-log10(FDR q-val)'].min())
    )
    dotplot_data['SizeNorm'] = dotplot_data['SizeNorm'] * (max_size - min_size) + min_size
    dotplot_data['alpha'] = np.where(dotplot_data['FDR q-val'] ==1, 0.3, 1.0)

    # Split data
    full_models = ['Sciplex', 'McFarland']
    tissue_models = ['Skin',
 'Urinary Tract',
 'Lung',
 'Pancreas',
 'CNS',
 'Large Intestine',
 'Kidney',
 'Breast',
 'Upper Aerodigestive Tract',
 'Thyroid',
 'Ovary',
 'Stomach',
 'Oesophagus',
 'Endometrium']
    treatment_models = ['Idasanutlin',
 'Trametinib',
 'Afatinib',
 'Dabrafenib',
 'Taselisib',
 'AZD5591',
 'Navitoclax',
 'JQ1',
 'Everolimus']
    data_full = dotplot_data[dotplot_data['model'].isin(full_models)].copy()
    data_tissue = dotplot_data[dotplot_data['model'].isin(tissue_models)].copy()
    data_treatment = dotplot_data[dotplot_data['model'].isin(treatment_models)].copy()

    # Plotting
    fig, axes = plt.subplots(1, 3, figsize=figure_size, dpi=300, sharey=True,
                            gridspec_kw={'width_ratios': [1, 10, 5]})


    for ax, sub_data, title in zip(
        axes,
        [data_full, data_tissue, data_treatment],
        ['Full models','Tissue-specific models','Drug-specific models' ]
    ):
        # X-axis model positions
        model_order = sorted(sub_data['model'].unique())
        model_to_x = {model: i for i, model in enumerate(model_order)}

        # Y-axis term positions (based on clustering)
        term_to_y = {term: i for i, term in enumerate(row_order)}
        ax.yaxis.grid(True, linestyle=':', linewidth=0.5, alpha=0.5)
        ax.set_axisbelow(True)  # Ensures grid is below all plot elements

        # Plot
        sc = ax.scatter(
            x=[model_to_x[m] for m in sub_data['model']],
            y=[term_to_y[t] for t in sub_data['Term']],
            # alpha=sub_data['alpha'],
            s=sub_data['SizeNorm'],
            c=sub_data['NES'],
            cmap='coolwarm',
            norm=norm,
            edgecolors='black'

        )

        # X-axis labels
        ax.set_xticks(list(model_to_x.values()))
        ax.set_xticklabels(model_order, rotation=45, ha='right', rotation_mode='anchor')
        ax.set_title(title)
        ax.set_xlim(-0.5, len(model_order) - 0.5)

    # Y-axis labels (on first plot only)
    axes[0].set_yticks(range(len(row_order)))
    axes[0].set_yticklabels(row_order)


    if legend:
        # Colorbar
        cbar_ax = fig.add_axes([1, 0.5, 0.04, 0.4])  # [left, bottom, width, height]
        cbar = plt.colorbar(sc, cax=cbar_ax)
        cbar.set_label("NES")

        # Size legend
        size_legend_vals = [0, 1, 2, 3]  # adjust as appropriate
        size_legend_sizes = [
            ((v - size_vals.min()) / (size_vals.max() - size_vals.min())) * (max_size - min_size) + min_size
            for v in size_legend_vals
        ]
        legend_elements = [
            Line2D([0], [0], marker='o', color='w', label=f'{10**(-v)}',
                markerfacecolor='gray', markersize=np.sqrt(s), markeredgecolor='black')
            for v, s in zip(size_legend_vals, size_legend_sizes)
        ]
        size_legend = ax.legend(handles=legend_elements, title='FDR')
        size_legend.set_bbox_to_anchor((1.05, 0.2))
        ax.add_artist(size_legend)

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.1)   # smaller → less whitespace

    return fig, axes

fig, axes = create_gsea_heatmap(dotplot_data)
fig.savefig(os.path.join(figures_dir, 'GSEA_dotplot.pdf'), bbox_inches='tight')
```

```python
def create_gsea_heatmap(dotplot_data, figure_size=(8, 4), legend=True):
    nes_matrix_all = dotplot_data.pivot(index="Term", columns="model", values="NES").fillna(0)

    # Try to identify the Observation column (flexible matching)
    obs_col = [c for c in nes_matrix_all.columns if 'Measured' in c]

    # Sort by the Observation NES values (descending)
    row_order = nes_matrix_all.sort_values(by=obs_col, ascending=False).index.tolist()

    from matplotlib.colors import PowerNorm

    # Color normalization
    max_nes = dotplot_data['NES'].abs().quantile(0.95)
    norm = PowerNorm(gamma=0.5, vmin=-max_nes, vmax=max_nes)

    # Normalize size
    min_size = 60
    max_size = 600
    size_vals = dotplot_data['-log10(FDR q-val)']
    dotplot_data['SizeNorm'] = (
        (dotplot_data['-log10(FDR q-val)'] - dotplot_data['-log10(FDR q-val)'].min()) /
        (dotplot_data['-log10(FDR q-val)'].max() - dotplot_data['-log10(FDR q-val)'].min())
    )
    dotplot_data['SizeNorm'] = dotplot_data['SizeNorm'] * (max_size - min_size) + min_size
    dotplot_data['alpha'] = np.where(dotplot_data['FDR q-val'] == 1, 0.3, 1.0)

    data_post = dotplot_data[dotplot_data['model'].str.contains('post')].copy()
    data_LFC = dotplot_data[dotplot_data['model'].str.contains('lfc')].copy()

    data_post['model'] = data_post['model'].str.replace('post ', '')
    data_LFC['model'] = data_LFC['model'].str.replace('lfc ', '')
    print(data_LFC['model'].unique())
    print(data_post['model'].unique())

    # Plotting
    fig, axes = plt.subplots(1, 2, figsize=figure_size, dpi=300, sharey=True,
                            gridspec_kw={'width_ratios': [7,6]})


    for ax, sub_data, title in zip(
        axes,
        [data_post,data_LFC],
        ['Trained on post- profiles','Trained on Ln(FC) profiles']
    ):
        # X-axis model positions
        if title == 'Trained on post- profiles':
            model_order = [
        'Measured','CPA',  'GEARS', 'GEARS\nOptimized', 'scFoundation', 'Average effect','No effect']
        else:
            model_order = [
        'Measured','CPA', 'GEARS', 'GEARS\nOptimized', 'scFoundation', 'Average effect']

              # If fewer than 2 models, skip clustering to avoid pdist error
                # --- Column clustering (keep 'Observation' fixed on the left) ---
        nes_matrix_sub = sub_data.pivot(index='Term', columns='model', values='NES').fillna(0)

        model_to_x = {model: i for i, model in enumerate(model_order)}

        # Y-axis term positions (based on clustering)
        term_to_y = {term: i for i, term in enumerate(row_order)}
        ax.yaxis.grid(True, linestyle=':', linewidth=0.5, alpha=0.5)
        ax.set_axisbelow(True)  # Ensures grid is below all plot elements

               # --- Compute NES sign agreement with Observation ---
        if 'Measured' in nes_matrix_sub.columns:
            obs_sign = np.sign(nes_matrix_sub['Measured']).replace(0, np.nan)  # sign per term
            # Map sign agreement for each (term, model)
            sub_data['sign_agree'] = [
                np.sign(row['NES']) == obs_sign.get(row['Term'], np.nan)
                for _, row in sub_data.iterrows()
            ]
        else:
            sub_data['sign_agree'] = False
        
        # --- Plot (thicker border if sign agrees with Observation) ---
        xs = [model_to_x[m] for m in sub_data['model']]
        ys = [term_to_y[t] for t in sub_data['Term']]

        # Base scatter (no agreement)
        sc = ax.scatter(
            x=xs,
            y=ys,
            s=sub_data['SizeNorm'],
            c=sub_data['NES'],
            cmap='coolwarm',
            norm=norm,
            edgecolors='black',
            alpha = np.where(sub_data['sign_agree'], 1, 0.3)
            # linewidths=np.where(sub_data['sign_agree'], 1.8, 0.5),  # thicker border if agreement
        )


        # X-axis labels
        ax.set_xticks(list(model_to_x.values()))
        ax.set_xticklabels(model_order, rotation=45, ha='right', rotation_mode='anchor')
        ax.set_title(title)
        ax.set_xlim(-0.5, len(model_order) - 0.5)

    # Y-axis labels (on first plot only)
    axes[0].set_yticks(range(len(row_order)))
    axes[0].set_yticklabels(row_order)

    if legend:
        # Colorbar
        cbar_ax = fig.add_axes([1, 0.5, 0.04, 0.4])  # [left, bottom, width, height]
        cbar = plt.colorbar(sc, cax=cbar_ax)
        cbar.set_label("NES")

        # Size legend
        size_legend_vals = [0, 1, 2]  # adjust as appropriate
        size_legend_sizes = [
            ((v - size_vals.min()) / (size_vals.max() - size_vals.min())) * (max_size - min_size) + min_size
            for v in size_legend_vals
        ]
        legend_elements = [
            Line2D([0], [0], marker='o', color='w', label=f'{10**(-v)}',
                markerfacecolor='gray', markersize=np.sqrt(s), markeredgecolor='black')
            for v, s in zip(size_legend_vals, size_legend_sizes)
        ]
        size_legend = ax.legend(handles=legend_elements, title='FDR', labelspacing=1.2, frameon=False)
        size_legend.set_bbox_to_anchor((1, 0.3))
        ax.add_artist(size_legend)

    plt.tight_layout()
    plt.subplots_adjust(wspace=0.1)   # smaller → less whitespace

    return fig, axes

fig, axes = create_gsea_heatmap(dotplot_data, figure_size=(9, 4))
fig.savefig(os.path.join(figures_dir, 'GSEA_dotplot_sciplex.pdf'), bbox_inches='tight')
```
